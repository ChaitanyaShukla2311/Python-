```python
#4.mini calculator using a dictionary of lambda functions
cal={'+':lambda a,b:a+b,
    '-':lambda a,b:a-b,
    '*':lambda a,b:a*b,
    '/':lambda a,b:a/b if b!=0 else print('Error')}
print(cal['+'](3,4))
print(cal['-'](3,4))
print(cal['*'](3,4))
print(cal['/'](3,4))
```

```python
#5.create a list of prime number using function
def is_prime(x):
    for j in range(2,x):
        if x%j==0:
            return False
    return True
        
prime=[i for i in range(2,101) if is_prime(i)]
print(prime)
```

```python
#6.most frequent character in a string
string='programmingisfununtilyouhaveaerror '
freq={}
for i in string:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
max1=max(freq,key=lambda x:freq[x])
print(freq)
print("most frequent character in a string is:",max1)
```

```python
#7.print letter 'A' in pattern
result_str="";    
for row in range(0,7):    
    for column in range(0,7):     
        if (((column == 1 or column == 5) and row != 0) or ((row == 0 or row == 3) and (column > 1 and column < 5))):    
            result_str=result_str+"*"    
        else:      
            result_str=result_str+" "    
    result_str=result_str+"\n"    
print(result_str)
```

```python
#8.print letter 'G' in pattern
result_str="";    
for row in range(0,7):    
    for column in range(0,7):     
        if ((column == 1 and row != 0 and row != 6) or ((row == 0 or row == 6) and column > 1 and column < 5) or (row == 3 and column > 2 and column < 6) or (column == 5 and row != 0 and row != 2 and row != 6)):  
            result_str=result_str+"*"    
        else:      
            result_str=result_str+" "    
    result_str=result_str+"\n"    
print(result_str);
```

```python
#9.Write a Python program to calculate a dog's age in dog's years. 
#Note: For the first two years, a dog year is equal to 10.5 human years. After that, each dog year equals 4 human years.
#Input a dog's age in human years: 15
#The dog's age in dog's years is 73

hyear=int(input("Input a dog's age in human years:"))
dogyear=10.5
if hyear<3:
    print("The dog's age in dog's years is",dogyear*hyear)
elif hyear>3:
    print("The dog's age in dog's years is",int((dogyear*2)+(hyear-2)*4))
```
