```python
def rev(string):
    print(string[::-1])
```

```python
def vowel_in(string):
    s=set()
    for ch in string:
        if ch in 'aeiouAEIOU':
            s.add(ch)
    return s
```

```python
def len_of(string):
    count=0
    for i in string:
        count+=1
    return count
```

```python
def is_palindrome(string):
    if string==string[::-1]:
        return True
    return False
    
```

```python
def to_lower(string):
    return string.lower()
    
```

```python
def to_upper(string):
    return string.upper()

    
```

```python
def word_in(string):
    return string.split()
    
```

```python
def to_ascii(string):
    return (ord(ch) for ch in string)
    
```
