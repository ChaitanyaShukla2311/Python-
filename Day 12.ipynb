```python
#4.replace words starting with 'a' or 'A' with @
sen='Apple at the start of the keeps doctors away'
words=sen.split()
new=[]
for i in words:
    if i[0]=='a' or i[0]=='A':
        new.append('@')
    else:
        new.append(i)
replaced=reduce(lambda x,y:x+" "+y,new)
replaced
```

```python
#8.replace spaces with uderscore
sen="datascience is fun"
res=""
for ch in sen:
    if ch==" ":
        res+='_'
    else:
        res+=ch
res
```

```python
#9.print all words starting and ending with same character
sen='hello madam how can i help you'
words=sen.split()
for i in words:
    if i[0]==i[-1]:
        print(i)
```

```python
#10.check if a sentence is pangram or not(contains all letters)
sen="The quick brown fox jumps over the lazy dog "
ch='abcdefghijklmnopqrstuvwxyz'
flag=True
for i in sen.lower():
    if i not in ch:
        flag==False
        break
print("pangram"if flag==True else "not pangram")
```
