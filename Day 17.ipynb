Programs based on exceptional handling with built in exceptions and custom also

```python
#1.zerodivisionerror and valueerror during division of two numbers
try:
    a=int(input("Enter numerator:"))
    b=int(input("Enter denominator:"))
    div=a/b
    print("division:",div)
except ZeroDivisionError:
    print("denominator cannot be zero")
except ValueError:
    print("Enter valid integer value")
finally:
    print("this is finally block")
```

```python
#2.IndexError when accessing the list
l=[1,2,3,4,5]
try:
    ind=int(input("Enter index"))
    print("Value:",l[ind])
except IndexError:
    print("Entered index is out of range")
```

```python
#3.ModuleNotFoundError when importing a package
try:
    import ahdn
except ModuleNotFoundError:
    print("invalide module name")
```

```python
#4.NameError when accessing a variable
try:
    print(d)
except NameError:
    print("name not defined")
```

```python
#5.KeyError when key is not in dictionary
d={'a':'Ram'}
try:
    key=input("Enter key")
    print("Value:",d[key])
except KeyError:
    print("key not in dictionary")
```

```python
#6.TypeError when giving invalid arguments 
try:
    def add(a,b):
        print(a+b)
    add(3,4,5)
except TypeError:
    print("give two arguments only")
```

```python
#7.AttributeError when name of function invalid
try:
    l=[]
    l.apend(4)
except AttributeError:
    print("give proper attribute name ")
```

```python
#8.custom exception for voting age 
class AgeException(Exception):
    pass
def checkage(age):
    if age<18:
        raise AgeException("Underage for voting")
    return "Eligible for voting"

try:
    age=int(input("Enter age "))
    print(checkage(age))
except AgeException as e:
    print("Error",e)
```

```python
#9.ValueError during string to integer conversion
def convert_to_int(s):
    try:
        return int(s) 
    except ValueError:
        print("characters cannot converted to int")

#print(convert_to_int('123'))
print(convert_to_int('abc'))
```

```python
#10.year of birth validator
try:
    year=int(input("Enter your birth year:"))
    if year<1900 or year>2025:
        raise ValueError("Year is out of valid range")
    print("Valid year")
except ValueError as v:
    print("Error",v)
             
```
