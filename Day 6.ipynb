```python
#1.Print hollow rectangle of stars
rows=int(input("Enter rows:"))
columns=int(input("Enter columns:"))
for i in range(rows):
    for j in range(columns):
        if(i==0 or i==(rows-1) or j==0 or j==(columns-1)):
            print("*",end=' ')
        else:
            print(" ",end=' ')
    print()
```

```python
#2.print unique and duplicated element in tuple
tuple=(1,1,2,3,3,4,4,5,6,6,7,8)
unique=[]
duplicate=[]
for i in tuple:
    if(tuple.count(i)==1):
        unique.append(i)
    else:
        if(i not in duplicate):
            duplicate.append(i)
print("unique elements:",unique)
print("duplicate elements:",duplicate)
    
```

```python
#3.create staircase pattern with numbers
n=int(input("Enter number of steps:"))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end='')
    print()
```

```python
#5.Merge two tuples element wise into a list of pairs
t1=(1,2,3)
t2=('A','B','C')
paired=[]
for i in range(len(t1)):
    paired.append((t1[i],t2[i]))
print(paired)

#tup=tuple(paired)
#tup
```

```python
#6.separate alphabets,digits and symbols from a string
string=input("Enter string:")
alpha=[]
digit=[]
symbol=[]
for i in string:
    if(i.isalpha()):
        alpha.append(i)
    elif(i.isdigit()):
        digit.append(i)
    else:
        symbol.append(i)
print("alphabets:",alpha,"\ndigits:",digit,"\nsymbols",symbol)
        
```

```python
#7.print all two digit number where sum of digits is even
for i in range(10,100):
    ten=i%10
    unit=i//10
    sum=ten+unit
    if(sum%2==0):
        print(i,end=' ')
```

```python
#8.print how many times a digit appears in list
l1=[23,38,18,372,36,48,47,27,8,17,26,17,173]
count=0
n=input("Enter digit to count:")
for i in l1:
    count=count+str(i).count(n)
print(n,"appears",count,"times")
```
