Programs based on Set and its methods

```python
#1.remove duplicate elements from list using set
l1=[1,2,4,3,4,1,2,2,5,5]
unique=list(set(l1))
print(unique)
```

```python
#2.find union and intersection of two sets 
a={1,2,3,4,5}
b={4,5,6,7,8}
print("union:",a|b)    #a.union(b)
print("intersection:",a&b)     #a.intersection(b)
```

```python
#3.add multiple elements to set
s={1,2,3}
s.update([4,5,6,7,8])
s
```

```python
#4.find common elements from two list
l1=[1,2,3,4,5,6,7]
l2=[3,4,6,9,0,1,5]
print(l1)
print(l2)
common=set(l1)&set(l2)
print("common :",common)
```

```python
#5.find who plays either cricket or football but not both
cricket={"john","dustin","mike","lucas","wil"}
football={"lucas","max","john","el"}
only=cricket^football
print("Play either cricket or football but not both:",only)
```

```python
#6.create a set of squares from 1 to 10 and filter even squares
squares=set()
for i in range(1,11):
    squares.add(i*i)
print("set of squares:",sorted(squares))
filtered=set()
for i in squares:
    if(i%2==0):
        filtered.add(i)
print("even squares :",sorted(filtered))
```

```python
#7.count unique words in a sentence
sen="there is a test this is only a test"
words=sen.split()
unique=set(words)
print("unique words:",unique)
print("count:",len(unique))
```

```python
#8.compare two list and print items that are not common 
list1=input("Enter list 1:").split()
list2=input("Enter list 2:").split()
print("items that are not common:",sorted(set(list1)^set(list2)))
```

```python
#9.check inputed sentence is of unique words or not
sen=input("Enter sentence:").lower().split()
if(len(sen)==len(set(sen))):
    print("Sentence is of all unique words")
else:
    print("Sentence is not of all unique words")
```

```python
#10.print characters which are repeated more than 2 times in a word
word=input("Enter a long word:")
repeat=set()
for i in word:
    if(word.count(i)>2):
        repeat.add(i)
print("characters which are repeated more than 2 times in a word:",repeat)
```
