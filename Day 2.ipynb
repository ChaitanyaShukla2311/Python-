Programs based on While loop

```python
#1.sum of first n natural numbers 
n=int(input("Enter n"))
i=0
sum=0
while(i<=n):
    sum=sum+i
    i=i+1
print("Sum of first",n,"natural numbers =",sum)
```

```python
#2.Print even numbers from 1 to 20
i=0
while(i<20):
    print(i+2)
    i=i+2
```

```python
#3.count digits in number 
num=int(input("Enter number :"))
count=0
i=0
while(num!=0):
    num=num//10
    count=count+1
print(count)
```

```python
#4.Reverse the numbner 
num=int(input("Enter number :"))
rev=0
rem=0
while(num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
print(rev)
```

```python
#5.find m^n without any inbuit function
base=int(input("ENter base"))
power=int(input("Enter power"))
i=1
value=1
while(i<=power):
    value=base*value
    i=i+1
print(base,"^",power,"=",value)
```

```python
#6.Find the factorial of number 
a=int(input("Enter number"))
i=1
fact=1
while(i<=a):
    fact=fact*i
    i=i+1
print("factorial =",fact)
```

```python
#7.Check number is palindrome or not
num=int(input("Enter number:"))
a=num
rem=0
rev=0
while(num>0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
print("Reverse: ",rev)
if(rev==a):
    print("Number is palindrome")
else:
    print("Number is not palindrom")
```

```python
#8.check number is armstrong or not (three digit only)
num=int(input("Enter number:"))
a=num
rem=0
arm=0
while(num>0):
    rem=num%10
    arm=arm+rem**3
    num=num//10
if(arm==a):
    print("Number is armstrong number")
else:
    print("Number is not armstrong number")
```
