```python
#2.inputed number is divisible by 5 or not
a=float(input("Enter number"))
if(a%5==0):
    print("Divisible by 5")
else:
    print("Not divisible by 5")
```

```python
#3.largest number between 3 inputed numbers
a=float(input("Enter no.1"))
b=float(input("Enter no.2"))
c=float(input("Enter no.3"))
if(a>b and a>c):
    print("no.1 is the largest number")
elif(b>a and b>c):
    print("no.2 is the largest number")
else:
    print("no.3 is the largest number")
    
              
```

```python
#4.simple grading based on inputed marks 
marks=int(input("Enter marks -"))
print("grades :")
if(35<marks<50):
    print("D")
elif(50<marks<70):
    print("C")
elif(70<marks<85):
    print("B")
elif(85<marks<=100):
    print("A")
elif(marks<0 or marks>100):
    print("invalid input")
else:
    print("Fail")
```

```python
#5.check number is three digit or not
num=int(input("Enter any number"))
if(99<num<1000):
    print("number is three digit")
else:
    print("number is not three digit")
```

```python
#6.check number is perfect square or not 
num=int(input("Enter number "))
sqrt=num**0.5
print(sqrt)
if(sqrt==int(sqrt)):
    print("Number is perfect square")
else:
    print("Not a perfect square")
    
```

```python
#7.profit or loss 
cost=float(input("Enter cost price:"))
sell=float(input("Enter selling price:"))
if(sell>cost):
    print("Profit")
elif(sell==cost):
    print("Equal")
else:
    print("Loss")
```

```python
#8.check character is vowel or consonant
char=input("Enter character :")
if(char=='a' or char=='e' or char=='i' or char=='o' or char=='u'):
    print("It is a vowel")
elif(char=='A' or char=='E' or char=='I' or char=='O' or char=='U'):
    print("It is a vowel")
else:
    print("It is a consonant")
```

```python
#9.calculate Simple interest
p=int(input("Enter Amount:"))
r=float(input("Enter percentage of rate:"))
t=float(input("Enter time in years :"))
SI=(p*r*t)/100
print("Simple interest =",SI)
```
