```python
#2.create a class calculator capable of finding square cube and square root of number
class calculator:
    def __init__(self,num):
        self.num=num
        print("NUmber:",num)
    def square(self):
        print(f"square:{self.num**2}")
    def cube(self):
         print(f"cube:{self.num**3}")
    def root(self):
         print(f"square root:{self.num**3}")

a=calculator(4)
a.square()
a.cube()
a.root()
```

```python
#3.create a class train with methods to book ticket,get status and get fare 
from random import randint
class train:
    def __init__(self,trainno):
        self.trainno=trainno
    def status(self):
        print(f"train no.{self.trainno} is running on time")
    def fare(self,fro,to):
        print(f"Fare for train no.{self.trainno} from {fro} to {to} is {randint(200,1000)}")

t=train(112233)
t.status()
t.fare('pune','mumbai')
```

```python
#4.create class student with name and marks of 3 sub and has methods to cal avg and print grade
class student:
    def __init__(self,sub1,sub2,sub3):
        self.sub1=sub1
        self.sub2=sub2
        self.sub3=sub3

    def avg(self):
        print(f"average of three sujects is:{(self.sub1+self.sub2+self.sub3)/3}")

    def grade(self):
        avge=(self.sub1+self.sub2+self.sub3)/3
        if avge>=90:
            return "A"
        elif avge>=80:
            return "B"
        elif avge>=70:
            return "C"
        elif avge>=60:
            return "D"

c=student(90,87,93)
c.avg()
c.grade()
```

```python
#5.create a class book with title,price and create multiple books and display those price under 500
class book:
    def __init__(self,title,price):
        self.title=title
        self.price=price
    def display(self):
        print(f"Book title:{self.title} and price :{self.price}")

books=[book("atomic habits",399),
       book("Rich dad,Poor dad",450),
       book("The Alchemist",550)]

for b in books:
    if b.price<500:
        b.display()
```

```python
#6.create a class circle with radius and method to cal area and circumference
class circle:
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r**2 
    def circumference(self):
        return 2*3.14*self.r

c=circle(4)
print(c.area())
print(c.circumference())
```

```python
#7.create class car with its info and method to cal mileage 
class car:
    def __init__(self,name,distance,fuelused):
        self.name=name
        self.distance=distance
        self.fuelused=fuelused

    def mileage(self):
        print(f"car name:{self.name} ,runned {self.distance}km and used {self.fuelused}ltr fuel \nmileage:{round(self.distance/self.fuelused,2)}")

c=car('maruti swift',20000,89)
c.mileage()
```

```python
#8.create class user and method to check correct password
class user:
    
    def __init__(self,user,pwd):
        self.user=user
        self.pwd=pwd
         
    def check(self,pas):
        return self.pwd==pas

u=user('cjs','12345')
passs=input("Enter password")
if u.check(passs):
    print("Correct password")
else:
    print("Wrong password")
```

```python
#10.create class and method to cal simple interest 
class SI:
    def __init__(self,p,r,t):
        self.p=p
        self.r=r
        self.t=t

    def simple_int(self):
        return (self.p*self.r*self.t)/100
a=SI(10000,5,2)
a.simple_int()
```
