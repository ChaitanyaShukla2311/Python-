```python
#1.sentence word into camelcase
def to_camelcase(a):
    words=a.split()
    return ''.join(w.capitalize()+" "for w in words)
    
sen='hello world'
to_camelcase(sen)
```

```python
#2.print hollow right angle triangle
n=10
for i in range(1,n+1):
    for j in range(1,i+1):
        if j==1 or j==i or i==n:
            print("*",end=' ')
        else:
            print(" ",end=' ')
    print()
        
```

```python
#3.sandglass star pattern
n=10
for i in range(n):
    print(" "*i + " *"*(n-i))
for i in range(1,n):
    print(" "*(n-i-1) + " *"*(i+1))
```

```python
#4.letter x in pattern 
n=11
for i in range(n):
    for j in range(n):
        if j==i or j==n-i-1:
            print('*',end='')
        else:
            print(' ',end='')
    print()
```

```python
#5.letter c in pattern
n=5
for i in range(n):
    for j in range(n):
        if((i!=1 or i!=n-1) and j==0) or (i==0 or i==n-1 and j>0):
            print(" *",end='')
        else:
            print(' ',end='')
    print()
```

```python
#6.letter B in pattern
n=10
for i in range(n):
    for j in range(n):
        if((i!=1 or i!=n-1) and j==0 or j==9) or (i==0 or i==n-1 or i==(n/2)-1 and j>0):
            print("*",end='')
        else:
            print(' ',end='')
    print()
```

```python
#7convert number to words(1-99)
unit=['','one','two','three','four','five','six','seven','eight','nine','ten']
tens=['','','twenty','thirty','forty','fifty','sixty','seventy','eighty','ninety']
special=['twenty','eleven','twelve','thirteen','fourteen','fifteen','sixteen','seventeen','eighteen','nineteen']
num=int(input("Enter number between 1-99"))
if (1<=num<=10):
    print(unit[num])
elif(11<=num<=20):
    print(special[num%10])
else:
    print(tens[num//10] +"-"+unit[num%10])
```

```python
#8.check if a password has at least one lowercase,one uppercase and one digit
pwd=input("Enter password:")
if (any(ch.isupper()for ch in pwd) and any(ch.islower()for ch in pwd) and any(ch.isdigit()for ch in pwd)):
    print("your password is accepted")
else:
    print("please enter password with at least one lowercase,one uppercase and one digit")
```

```python
#9.check if email contains @ and ends with .com
email=input("Enter your email:")
if ('@'in email and email.endswith('.com')):
    print("valid email")
else:
    print("Invalid email")
```
