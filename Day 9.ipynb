```python
#1.function to find maximum of three numbers
def maximum(n1,n2,n3):
    if(n1>n2 and n1>n3):
        print("maximum:",n1)
    elif(n2>n1 and n2>n3):
        print("maximum:",n2)
    else:
        print("maximum:",n3)

maximum(10,20,15)
```

```python
#2.function to add all elements in list
def total(l1):
    count=0
    for i in l1:
        count+=i
    return count

num=[1,2,3,4,5,6,7,8]
total(num)
```

```python
#3.function to reverse a string
def rev(a):
    return a[::-1]
name='chaitanya'
rev(name)
```

```python
#4.function to find factorial of that number
def fact(n1):
    fact=1
    for i in range(1,n1+1):
        fact*=i
    return fact

fact(5)
```

```python
#5.function to count words in a sentence
def words(a):
    b=a.split()
    return len(b)

sen="hello i am a python learner"
words(sen)
```

```python
#6function that checks whether a passed string is a palindrome or not.
def palindrome(a):
    if(a==a[::-1]):
        return "true"
    else:
        return "false"

a='nayan'
palindrome(a)
```

```python
#7.function to check if a number is prime 
def isprime(num):
    while(num>1):
        for i in range(2,num):
            if(num%i==0):
                return False
            return True

isprime(17)
```

```python
#9.funtion to print multiplication table
def table(n):
    for i in range(1,11):
        print(n,"*",i,"=",n*i)
table(3)
```

```python
#10.function to find index of number in list
def index(list,n):
    for i in range(len(list)):
        if(list[i]==n):
           print(i,end=' ')

l1=[1,4,6,7,3,9,0]
index(l1,7)
```
