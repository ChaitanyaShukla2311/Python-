Programs based on for loop 

```python
#1.Print numbers from 10 to 1 
for i in range(10,0,-1):
    print(i)
```

```python
#2.Print square of each number from 1 to 10
for i in range(1,11):
    print(i,"^",i,"=",i*i)
```

```python
#3.print reverse of the string
a=str(input("Enter string"))
l=len(a)
for i in range(l-1,-1,-1):
        print(a[i],end='')
```

```python
#5.print prime numbers between 1 and 50
for i in range(2,51):
    prime=True
    for j in range(2,i):
        if(i%j==0):
            prime=False
            break
    if(prime==True):
        print(i,end=' ')
```

```python
#6.Find all factors of a number
num=int(input("Enter number"))
print("Factors of number")
for i in range(1,num+1):
    if(num%i==0):
        print(i,end=' ')
```

```python
#7.inverted triangle pattern
for i in range(4,0,-1):
    print("*"*i)
```

```python
#8.pyramid of *
for i in range(1,10):
    print(" "*(10-i)+"* "*i)
```

```python
#9.diamond pattern
for i in range(1,4):
    print(" "*(4-i)+"* "*i)
for j in range(2,0,-1):
    print(" "*(4-j)+"* "*j)
```
