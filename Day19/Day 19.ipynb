Programs based on I/O file handling

```python
#1.write to a file
with open("sample.txt","w") as f:
    f.write("this is written by write()")
    
```

```python
#2.Read from a file
with open("sample.txt") as f:
    content=f.read()
    print(content)
f.close()
```

```python
#3.append to an existing file
with open("sample.txt","a") as f:
    f.write("\nthis is added later")
f.close()
```

```python
#4.read line by line 
with open("sample.txt") as f:
    for line in f:
        print(line.strip())
f.close()
```

```python
#5.create file and write 5 lines into it
with open("demo.txt","w") as d:
    for i in range(5):
        d.write(f"this is line number {i}")
d.close()
```

```python
#6.copy content from one file to another
with open("sample.txt","r") as s,open("demo.txt","a") as d:
    d.write(s.read())
s.close()
d.close()
```

```python
#8.generate tables from 2 to 10 in different files

def generate_tables(n):
    table=""
    for i in range(1,11):
        table += f"{n} x {i} ={n*i}\n"

    with open(f"tables/table_{n}.txt","w") as f:
        f.write(table)

for i in range(2,11):
    generate_tables(i)
    
```

```python
#9.a file contains a word "donkey" multiple times you need to replace this word with ###### by updating the same file
word='donkey'

with open("file.txt") as f:
    content=f.read()

newcontent=content.replace(word,"######")

with open("file.txt","w") as f:
    f.write(newcontent)
```

```python
#10.write a program to find out where the word python is present on which line 

with open("text.txt") as f:
    lines=f.readline()
lineno=1
for line in lines:
    if('Python' in lines):
        print(f"Yes python is present on line no {lineno}")
        break
    lineno +=1
else:
    print("Not present")

f.close()
```
