programs based on lambda function

```python
#1.find maximum of two numbers
maximum=lambda a,b:a if a>b else b
maximum(10,5)
```

```python
#2.check number is even
even=lambda a:a%2==0
even(8)
```

```python
#3.convert celcius to fahrenheit
to_far=lambda c:(9/5)*c+32
print(to_far(20))
```

```python
#4.check number is positive,negitive
check=lambda x:"positive" if x>0 else "negative"
check(-10)
```

```python
#5.check vowel or not
is_vowel=lambda x:x in 'aeiouAEIOU'
is_vowel('b')
```

```python
#6.check leap year or not
is_leap=lambda x:x%4==0 and (x%100!=0 or x%400==0)
is_leap(2024)
```

```python
#7.ckeck fail or pass 
grade = lambda score:"Pass" if score>=50 else "Fail"
print(grade(67))
```

```python
#8.check two strings are anagram or not
is_anagram=lambda s1,s2:sorted(s1)==sorted(s2)
is_anagram('listen','silent')
```

```python
#9.cal square root of number
sqrroot=lambda x:round(x**0.5,2)
sqrroot(10)
```

```python
#10.check list is palindrome or not
is_palindrome=lambda l1:l1==l1[::-1]
list1=[1,2,3,2,1]
is_palindrome(list1)
```
