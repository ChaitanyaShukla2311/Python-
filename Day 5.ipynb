```python
#3.minimum and maximum value in tuple with built in fuctions
num=(4,1,0,9,7,3)
maximum=num[0]
minimum=num[0]
for i in num:
    if(i>maximum):
        maximum=i
    if(i<minimum):
        minimum=i
print("maximum:",maximum,"  minimum:",minimum)

```

```python
#4.extract even number from given tuple
given=(7,2,8,3,4,1,6)
even=[]
for i in given:
    if(i%2==0):
        even.append(i)
print("Even numbers :",even)
```

```python
#6.find length of longest string in tuple
tuple=("ramesh","sujal","ram","mohit","chaitanya")
a=0
for i in tuple:
    if(len(i)>a):
        a=len(i)
print(a)
```

```python
#7.find pairs with a given sum
tuple=(1,2,3,4,5,6)
sum=7
pairs=[]
for i in range(len(tuple)):
    for j in range(len(tuple)):
        if(tuple[i]+tuple[j]==sum):
            pairs.append((tuple[i],tuple[j]))
print(pairs)    
```

```python
#8.filter only strings from tuple
t=(1,'abc',4,'xyz',True)
Str=[]
for i in range(len(t)):
    if(type(t[i])==str):
        Str.append(t[i])
Str
```
