Programs based on dictionary 

```python
#1.create a dictionary and print all keys and value separately
d1={1:'a',2:'b',3:'c',4:'d',5:'e'}
print(d1.keys())
print(d1.values())
print(d1.items())
```

```python
#2.create a dictionary of 5students with their marks and print who scored above 85
stud={'ram':90,'sham':93,'om':87,'sai':79,'shiv':84}

for name,mark in stud.items():
    if(mark>85):
        print(name,"scored",mark)
```

```python
#3.find a key with maximum value 
data ={'a':10,'b':50,'c':30,'d':20}
maxkey=max(data.values())
maxkey
for i,j in data.items():
    if(j==maxkey):
        print("key with maximum value is",i,"with value",j)
```

```python
#5.create a dictionary of words and its lengths in a sentence
sen="practice makes man pefect"
words=sen.split()
d1={}
for i in words:
    d1.update({i:len(i)})
d1   
```

```python
#6.convert two list into a dictionary with keys as first list and values as second
l1=['name','age','city']
l2=['chaitanya',20,'Pune']

d1=dict(zip(l1,l2))
d1
```

```python
#7.remove all values from dict less than 50
d1={'a':10,'b':50,'c':70,'d':20,'e':55,'f':28,'g':60}
l1=[]
for i,j in d1.items():
    if(j<50):
        l1.append(i)
for i in l1:
    d1.pop(i)
d1
```

```python
#8.find common keys in two dict
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'b': 20, 'c': 30, 'd': 40}

common=set(dict1.keys()) & set(dict2.keys())
common
```

```python
#9.create a dict of cubes of 1 to 10 
cube={}
for i in range(1,11):
    cube.update({i:i*i})
cube
```
