Programs based on exceptional handling including class

```python
#1.Bank acc with withdrawl exception
class bankacc:
    def __init__(self,bal):
        self.bal=bal

    def withdraw(self,amt):
        if amt>self.bal:
            raise Exception("Insuffucient balance")
        self.bal-=amt
        return f"Transaction Successful, balance:{self.bal}"

try:
    acc=bankacc(10000)
    acc.withdraw(15000)
except Exception as e:
    print("Transaction Failed\n",e)
```

```python
#2.product availability checker
class product:
    def __init__(self,p,qty):
        self.p=p
        self.qty=qty

    def is_available(self):
        if self.qty<1:
            raise ValueError("Product out of stock")
        return f"Product:{self.p} is available"

try:
    pen=product('Pen',0)
    print(pen.is_available())
except ValueError as v:
    print("Inventory Error\n",v)
```

```python
#3.mobile number validator with custom exception
class InvalidMobileError(Exception):
    pass

class number:
    def __init__(self,name,no):
        self.name=name
        self.no=no

    def is_valid(self):
        if not(self.no.isdigit() and len(self.no)==10):
            raise InvalidMobileError("Invalid Mobile Number")
        return "VALID MOBILE NUMBER"

try:
    ram=number('Ram','9872619k11')
    print(ram.is_valid())
except InvalidMobileError as m:
    print("Error",m)
```

```python
#4.parcel weight limit checker
class OverLimitError(Exception):
    pass

class parcel:
    def __init__(self,weight):
        self.weight=weight
    
    def checkweight(self):
        if self.weight>25:
            raise OverLimitError("Weight is out of limit")
        return "Pass"

try:
    p=parcel(30)
    print(p.checkweight())
except OverLimitError as o:
    print("Overweight\n",o)
```

```python
#5.library book return limit checker
class LateError(Exception):
    pass
class library:
    def __init__(self,book,days):
        self.book=book
        self.days=days
    def check(self):
        if self.days>15:
            raise LateError("Book returned late!!!")
        return "Book return on time"

try:
    l=library('Atomic habit',18)
    l.check()
except LateError as l:
    print("!!!",l)
```

```python
#6.Car fuel limit checker
class FuelError(Exception):
    pass
class car:
    def __init__(self,fuel):
        self.fuel=fuel
    def check(self):
        if self.fuel<=0:
            raise FuelError("LOW FUEL!!")
        return "You can start car"
try:
    c=car(0)
    c.check()
except FuelError as f:
    print(f,"You can not start the car")
```

```python
#7.Electricity bill validator
class bill:
    def __init__(self,unit):
        self.unit=unit
    def valid(self):
        if self.unit<=0:
            raise ValueError("Units cannot be negative")
        return "valid"
try:
    b=bill(0)
    b.valid()
except ValueError as v:
    print("Error",v)
```

```python
#8.Exam score validator 
class InvalidScoreError(Exception):
    pass
class exam:
    def __init__(self,score):
        self.score=score
    def check(self):
        if 0<=self.score>=100:
            raise InvalidScoreError("Score can not be more than 100 or negative")
        return self.score

try:
    s=exam(101)
    s.check()
except InvalidScoreError as i:
    print("Invalid score:",i)
```

```python
#9.Discount code validator 
class InvalidCodeError(Exception):
    pass
class cart:
    def __init__(self,code):
        self.code=code
    def apply_discount(self):
        valid=['ABC100','RED500']
        if self.code not in valid:
            raise InvalidCodeError("Invalid code")
        return "discount applied"
try:
    c=cart('ABC200')
    c.apply_discount()
except InvalidCodeError as d:
    print("Can not apply discount",d)
```

```python
#10.null pointer exception
class student:
    def __init__(self,graduation):
        self.graduation=graduation
    def is_graduate(self):
        if self.graduation=='no' :
            raise TypeError("Student not eligible")
        return True

try:
    s=student('no')
    s.is_graduate()
except TypeError as t:
    print("Must be graduate",t)
```
