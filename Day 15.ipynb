```python
#1.autocorrect suggestion,input a mistyped word and return its closest match from list of known words
dictionary=['hello','world','python','program','developer','data','science']

def similar(w1,w2):
    length_diff=abs(len(w1)-len(w2))
    common=sum(1 for a,b in zip(w1,w2) if a!=b)
    return common+length_diff

def autocorrect(word,dictionary):
    word=word.lower()
    if word in dictionary:
        return word
    correct=[(similar(word,d),d) for d in dictionary]
    correct.sort()
    return correct[0][1]

w=input("Enter a word")
corrected=autocorrect(w,dictionary)
print("did you mean:",corrected)
```

```python
#5.Generate acronym from phrase
text=input("Enter your phrase:")
words=text.split()
acro="".join(w[0].upper() for w in words if w[0].isupper())
print(acro)
```

```python
#6.convert number to words in a sentence
nums={'0':'zero','1':'one','2':'two','3':'three','4':'four','5':'five','6':'six','7':'seven','8':'eight','9':'nine'}
text=input("Enter sentence with numbers:")
words=text.split()
converted=[]
for w in words:
    if w.isdigit():
        towords=''.join(nums[d] for d in w)
        converted.append(towords)
    else:
        converted.append(w)
print(" ".join(converted))
```

```python
#7.Encrypt text with caesar cipher 
text=input("Enter lowercase text:")
encrypted=""
for ch in text:
    if 'a'<=ch<='z':
        cipher=chr(((ord(ch)-ord('a')+3)%26)+ord('a'))
        encrypted+=cipher
    else:
        encrypted+=ch
print("Encypted",encrypted)
```

```python
#8.detect negative review 
review=input("Enter a review:").lower()
neg=['bad','worst','poor','horrible']
found=any(word in review.split() for word in neg)
print("Negative review" if found else "Seems fine")
```
